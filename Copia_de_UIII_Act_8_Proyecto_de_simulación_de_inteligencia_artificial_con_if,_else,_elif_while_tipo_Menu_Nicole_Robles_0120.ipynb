#### Problema Periodista y libreta

```python
# --- SIMULADOR DE IA: ASISTENTE DE PERIODISTA Y LIBRETA ---

# Créditos iniciales (se ejecutan una sola vez fuera del ciclo)
print("Programa realizado por Nicole Robles NL = 50")
print("Bienvenido al Cerebro Lógico del Periodista Digital")

# Variable de control para el menú
opcion_menu = "0"

# Estructura repetitiva para mantener el sistema activo
while opcion_menu != "3":
    print("\n--- MENÚ DE IA PERIODÍSTICA ---")
    print("1. Evaluar Noticia (Asistente de Redacción)")
    print("2. Gestionar Libreta (Organizador de Notas)")
    print("3. Apagar Sistema")

    opcion_menu = input("\nSeleccione una opción (1-3): ")

    # MODO 1: Evaluación inteligente de noticias
    if opcion_menu == "1":
        print("\n[MODO IA: EVALUADOR DE NOTICIAS]")
        titulo = input("Tema de la noticia: ")
        prioridad = int(input("Nivel de importancia (1 al 10): "))
        tiempo_horas = int(input("Horas transcurridas desde el evento: "))
        fuente_confiable = input("¿La fuente es verificada? (si/no): ").lower()

        # Lógica de IA para decidir el destino de la noticia
        if prioridad >= 8 and tiempo_horas <= 1 and fuente_confiable == "si":
            # Operador lógico 'and' para noticias de alto impacto
            print(f"IA: ¡ALERTA! '{titulo}' es una NOTICIA DE ÚLTIMA HORA.")
            print("IA: Recomendación: Publicar inmediatamente en redes sociales.")
        elif prioridad > 5 or fuente_confiable == "si":
            # Operador 'or' para noticias relevantes pero no urgentes
            print(f"IA: '{titulo}' es una Noticia Estándar.")
            print("IA: Recomendación: Escribir nota completa para el diario de mañana.")
        else:
            print(f"IA: '{titulo}' tiene baja relevancia actual.")
            print("IA: Recomendación: Guardar como antecedente en la base de datos.")

    # MODO 2: Organización inteligente de la libreta
    elif opcion_menu == "2":
        print("\n[MODO IA: ORGANIZADOR DE LIBRETA]")
        cantidad_paginas = int(input("¿Cuántas páginas de notas tienes?: "))
        bateria_dispositivo = float(input("Nivel de batería de tu libreta digital (0.0 a 100.0): "))
        es_entrevista = input("¿Son notas de una 'entrevista' o 'investigacion'?: ").lower()

        # Lógica para decidir el formato de almacenamiento
        if es_entrevista == "entrevista" and bateria_dispositivo > 20.0:
            print("IA: Perfil detectado: ENTREVISTA DINÁMICA.")
            # Operación aritmética simple para estimar tiempo de transcripción
            tiempo_estimado = cantidad_paginas * 15
            print(f"IA: Guardar en 'Libreta Digital'. Tiempo de transcripción: {tiempo_estimado} min.")
        elif cantidad_paginas > 10 or bateria_dispositivo < 10.0:
            # Si hay poca batería o muchas páginas, mejor usar papel
            print("IA: Perfil detectado: INVESTIGACIÓN LARGA.")
            print("IA: Recomendación: Usar LIBRETA FÍSICA (Papel) para evitar pérdida de datos.")
        else:
            print("IA: Perfil detectado: NOTA RÁPIDA.")
            print("IA: Recomendación: Guardar en el 'Bloc de Notas' del celular.")

    # MODO 3: Salida
    elif opcion_menu == "3":
        print("\nCerrando sesión de IA... ¡Buen trabajo, periodista!")

    # Manejo de opciones no válidas para alumnos principiantes
    else:
        print("\nIA: Error. Opción no reconocida por el sistema.")

print("\n--- Sistema Finalizado. Sesión Cerrada ---")
print("Programa realizado por Nicole Robles NL = 50")
```
